```python
class Employee:
  def __init__(self):
    self.name = "Rahul"   # Public
    self._salary = 500000  # Procted
    self.__password = "12345" # Private

  def show_details(self):
    print(self.name)
    print(self._salary)
    print(self.__password)

e = Employee()
print(e.name)             # Works
print(e._salary)          # Works, but discouraged
# print(e.__password)     # Error
e.show_details()
```

```python
class Employee:
  def __init__(self):
    self.name = "Anu"   # Public
    self._salary = 500000  # Procted
    self.__password = "12345" # Private

  def show_parent(self):
    print(self.name)
    print(self._salary)
    print(self.__password)

class Manager(Employee):
  def show_manager(self):
    print("Name :", self.name)           # Public ->accessible
    print("Salary :",self._salary)       # Protected ->accessible
    # print("Password :", self.__password)             # Private -> Not accessible

manager=Manager()
manager.show_manager()
```
